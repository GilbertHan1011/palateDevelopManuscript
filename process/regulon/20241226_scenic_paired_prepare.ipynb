```python
rna = sc.read_h5ad("../process/framework/obj/20241220_scenicplus_prepare_rna_matched.h5ad")
```

```python
# Create mapping dictionary for prefix replacements
prefix_map = {
    'C1_': 'E125_1:',
    'C2_': 'E125_2:',
    'B1_': 'E145_1:',
    'B2_': 'E145_2:'
}

# Function to replace prefixes
def replace_prefix(barcode):
    for old_prefix, new_prefix in prefix_map.items():
        if barcode.startswith(old_prefix):
            return barcode.replace(old_prefix, new_prefix)
    return barcode
new_barcodes = [replace_prefix(x) for x in rna.obs_names]
```

```python
def transform_barcode(barcode):
    # Split by ':' and get the components
    sample, barcode_seq = barcode.split(':')
    # Create new format
    return f"{barcode_seq}-{sample}___{sample}"

# Apply to list of barcodes
new_barcodes2 = [transform_barcode(x) for x in new_barcodes]
```

```python
rna.obs_names = new_barcodes2
```

```python
rna.write_h5ad("../process/framework/obj/20241220_scenicplus_prepare_rna_matched.h5ad")
```
