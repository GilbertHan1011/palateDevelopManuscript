```python
def transform_barcode_2(barcode):
    # Define prefix mapping
    prefix_map = {
        'B1_': 'E145_1',
        'B2_': 'E145_2',
        'C1_': 'E125_1',
        'C2_': 'E125_2'
    }
    
    # Split prefix and barcode sequence
    for old_prefix, new_prefix in prefix_map.items():
        if barcode.startswith(old_prefix):
            barcode_seq = barcode[len(old_prefix):]  # Remove old prefix
            return f"{barcode_seq}-{new_prefix}___{new_prefix}"
    
    return barcode  # Return original if no matching prefix

new_barcodes2 = [transform_barcode_2(x) for x in peakvi_umap.index]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

def plot_comparison(df1_series, df2_series, title="K14(-) Comparison"):
    # Get common indices
    common_indices = list(set(df1_series.index) & set(df2_series.index))
    
    # Create DataFrame with matched values
    comparison_df = pd.DataFrame({
        'Dataset1': df1_series[common_indices],
        'Dataset2': df2_series[common_indices],
        'TF': common_indices
    })
    
    # Create scatter plot
    plt.figure(figsize=(10, 8))
    
    # Basic scatter plot
    sns.scatterplot(data=comparison_df, x='Dataset1', y='Dataset2', alpha=0.6)
    
    # Add diagonal line
    min_val = min(comparison_df['Dataset1'].min(), comparison_df['Dataset2'].min())
    max_val = max(comparison_df['Dataset1'].max(), comparison_df['Dataset2'].max())
    plt.plot([min_val, max_val], [min_val, max_val], 'r--', alpha=0.5)
    
    # Calculate correlation
    corr = np.corrcoef(comparison_df['Dataset1'], comparison_df['Dataset2'])[0,1]
    
    # Add labels and title
    plt.xlabel('Paired K14(-) values')
    plt.ylabel('Unpaired 2 K14(-) values')
    plt.title(f'{title}\nCorrelation: {corr:.3f}')
    
    # Add annotations for outliers
    def annotate_outliers(df, threshold=2):
        z_scores = np.abs((df['Dataset1'] - df['Dataset2']) / 
                         (df['Dataset1'] - df['Dataset2']).std())
        outliers = df[z_scores > threshold]
        for idx, row in outliers.iterrows():
            plt.annotate(row['TF'], 
                        (row['Dataset1'], row['Dataset2']),
                        xytext=(5, 5), textcoords='offset points',
                        fontsize=8)
    
    annotate_outliers(comparison_df)
    
    plt.tight_layout()
    plt.savefig("../result/regulation/20241226_scenicplus_paired/20241228_K14_paired.pdf")
    plt.show()
    
    return comparison_df
    
df1 = rss_transpose.replace('NaN', np.nan).dropna()
df2 = rss_origin.replace('NaN', np.nan).dropna()
# Use the function
comparison_df = plot_comparison(df1.loc[:,"K14(-)"], df2.loc[:,"K14(-)"])

# Print summary statistics
print("\nSummary of differences:")
comparison_df['Difference'] = comparison_df['Dataset1'] - comparison_df['Dataset2']
print(comparison_df['Difference'].describe())

# Print most different TFs
print("\nLargest differences:")
print(comparison_df.nlargest(5, 'Difference')[['TF', 'Dataset1', 'Dataset2', 'Difference']])
print("\nSmallest differences:")
print(comparison_df.nsmallest(5, 'Difference')[['TF', 'Dataset1', 'Dataset2', 'Difference']])
```

```python
# Method 1: Remove columns with sum=0 and sort
def clean_and_sort_df(df):
    # Remove columns where sum is 0
    df = df.loc[:, (df != 0).any(axis=0)]
    
    # Sort columns by sum in descending order
    column_sums = df.sum()
    sorted_columns = column_sums.sort_values(ascending=False).index
    df = df[sorted_columns]
    
    return df
```
